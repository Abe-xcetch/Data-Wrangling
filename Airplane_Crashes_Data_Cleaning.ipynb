```python
# Qusetion 5: Create the fatality_locations DataFrame

fatality_locations = df[["Date", "Location", "Aboard", "Fatalities"]].copy()
```

The aboard and fatalities columns show that the no. of people aboard and the no. of fatalities for each recorded crash. There are 58 recorded crashes where the no. of fatalities is zero.

```python
# Question 8: Split location into region and state/country

fatality_locations[["Region", "State/Country"]] = fatality_locations["Location"].str.split(", ",n=1,  expand=True)
```

```python
fatality_locations['Region'] = fatality_locations['Region'].str.strip()
fatality_locations['State/Country'] = fatality_locations['State/Country'].str.strip()
```

```python
fatality_locations.head()
```

```python
# Question 9: Top 100 crashes by number of fatalities

top_100_fatalities = fatality_locations.sort_values(
    by="Fatalities", ascending=False
    ).head(100)

```

```python
top_100_fatalities
```

```python
top_100_fatalities.shape
```

```python
# Question 10: Distribution of the top 25 fatalities by country/ U.S. State

country_fatalities = (
    fatality_locations.groupby("State/Country")["Fatalities"]
    .sum()
    .sort_values(ascending=False)
    .head(25)
)

country_fatalities
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.barplot(x=country_fatalities.index, y=country_fatalities.values)
```
